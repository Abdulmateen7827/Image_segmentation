```python
import os
from PIL import Image
from pillow_heif import register_heif_opener

# Register HEIF support with PIL
register_heif_opener()

# Source and destination root folders
SRC_ROOT = '/Users/abdulmateen/Downloads/vegetables'        # your original folder
DST_ROOT = '/Users/abdulmateen/Downloads/vegetables1'   # where converted .heic files will go

# Ensure destination root exists
os.makedirs(DST_ROOT, exist_ok=True)

# Walk through class folders
for class_name in sorted(os.listdir(SRC_ROOT)):
    class_dir = os.path.join(SRC_ROOT, class_name)
    if not os.path.isdir(class_dir):
        continue

    dst_class_dir = os.path.join(DST_ROOT, class_name)
    os.makedirs(dst_class_dir, exist_ok=True)

    for fname in os.listdir(class_dir):
        if not fname.lower().endswith(".jpg"):
            continue

        src_path = os.path.join(class_dir, fname)
        dst_path = os.path.join(dst_class_dir, os.path.splitext(fname)[0] + '.heic')

        try:
            img = Image.open(src_path).convert("RGB")
            img.save(dst_path, format="HEIC", quality=90)
            print(f"✅ Converted: {src_path} → {dst_path}")
        except Exception as e:
            print(f"❌ Failed to convert {src_path}: {e}")
```
